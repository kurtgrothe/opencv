## Splitting and Merging Channels

```python
img_bgr = cv2.imread('../data/Emerald_Lakes_New_Zealand.jpg', cv2.IMREAD_COLOR)
b, g, r =  cv2.split(img_bgr)

# Dispaly the Channels
plt.figure(figsize=[20,10])
plt.subplot(141); plt.imshow(r,cmap='grey'); plt.title('Red Channel')
plt.subplot(142); plt.imshow(g,cmap='grey'); plt.title('Green Channel')
plt.subplot(143); plt.imshow(b,cmap='grey'); plt.title('Blue Channel')
```

## Color Spaces
**Organization of all human-perceivable colors that contain color models (RGB, HSV..etc) that form a mathematical representation**

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)

# Split into HSV Channels
h, s, v = cv2.split(img_hsv)

# Dispaly the Channels
plt.figure(figsize=[20,5])
plt.subplot(141); plt.imshow(h,cmap='grey'); plt.title('Hue Channel')
plt.subplot(142); plt.imshow(s, cmap='grey'); plt.title('Saturation Channel')
plt.subplot(143); plt.imshow(v, cmap='grey'); plt.title('Value Channel')

# Original Image in RBG, need to manipulate the channels
img_rgb = img_bgr[:,:,::-1]
img_hsv = cv2.merge([v,s,h])
plt.subplot(144); plt.imshow(img_hsv); plt.title('Original Image')
```

plt_imshow_hsv_art

```python
def plt_imshow_hsv_art(hsv_img):
    """Converts displays an HSV image with Matlotlib that's intended to use RGB format.
        Instead of RGB channels the function mixes HSV channels and displays each permutation."""
    
    h,s,v = cv2.split(hsv_img)
    
    # Create and display the permutations
    permutations = [(h, s, v, 'HSV'), (h, v, s, 'HVS'), 
                    (s, h, v, 'SHV'), (s, v, h, 'SVH'), 
                    (v, h, s, 'VHS'), (v, s, h, 'VSH')]
    
    plt.figure(figsize=[20,10])
    for i, (c1, c2, c3, title) in enumerate(permutations, start=1):
        merged_hsv = cv2.merge([c1, c2, c3])
        rgb_img = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2RGB)  # Convert to RGB for matplotlib
        plt.subplot(2, 3, i)
        plt.imshow(rgb_img)
        plt.title(title)
```

**pixel_id**

```python
def pixel_id(image,y_coord, x_coord):
    """Returns the pixel value and visual identification at a given coordinate"""
    plt.figure(figsize=[5,5])
    plt.imshow(image, cmap='gray')
    plt.axhline(y=y_coord, color='r', linewidth='1')
    plt.axvline(x=x_coord, color='r', linewidth='1')
    plt.title(f"Pixel Value: {image[y_coord, x_coord]}")
    
```

def change_pixel

```python
def change_pixel(image, y_coord, x_coord, new_value):
    """Changes the pixel value at a given coordinate"""
    modified = image.copy()
    modified[y_coord, x_coord] = new_value
    plt.figure(figsize=[10,5])

    # Original Image
    ax1 = plt.subplot(121)
    ax1.imshow(image, cmap='gray')
    ax1.set_title(f"Original \nPixel Value: {image[y_coord, x_coord]}, Coordinate: ({x_coord}, {y_coord})")   
    # Set the limits to image dimensions
    ax1.set_xlim([0, image.shape[1]])
    ax1.set_ylim([image.shape[0], 0])  # inverted y-axis for image coordinates
    # Draw lines that stop before the point
    ax1.plot([0, x_coord-1], [y_coord, y_coord], 'r', linewidth=1, alpha=0.5)
    ax1.plot([x_coord+1, image.shape[1]], [y_coord, y_coord], 'r', linewidth=1, alpha=0.5)
    ax1.plot([x_coord, x_coord], [0, y_coord-1], 'r', linewidth=1, alpha=0.5)  # Vertical line top part
    ax1.plot([x_coord, x_coord], [y_coord+1, image.shape[0]], 'r', linewidth=1, alpha=0.5)  # Vertical line bottom part

    # Modified Image
    ax2 = plt.subplot(122)
    ax2.imshow(modified, cmap='gray')
    ax2.set_title(f"Modified\nPixel Value: {modified[y_coord, x_coord]}, Coordinate: ({x_coord}, {y_coord})")
    # Set the limits to image dimensions
    ax1.set_xlim([0, image.shape[1]])
    ax1.set_ylim([image.shape[0], 0])  # inverted y-axis for image coordinates
    # Draw lines that stop before the point
    ax2.plot([0, x_coord-1], [y_coord, y_coord], 'r', linewidth=1, alpha=0.5)
    ax2.plot([x_coord+1, image.shape[1]], [y_coord, y_coord], 'r', linewidth=1, alpha=0.5)
    ax2.plot([x_coord, x_coord], [0, y_coord-1], 'r', linewidth=1, alpha=0.5)  # Vertical line top part
    ax2.plot([x_coord, x_coord], [y_coord+1, image.shape[0]], 'r', linewidth=1, alpha=0.5)  # Vertical line bottom part

    plt.show()



    
```

## Resize maintaining aspect ration

```python
width = 300
aspect_ratio = cropped_eagle.shape[1] / cropped_eagle.shape[0]
height = int(width * aspect_ratio)
dim = (width, height)

resized_cropped_region = cv2.resize(cropped_eagle, dim, interpolation = cv2.INTER_AREA)
plt.imshow(resized_cropped_region[:,:,::-1]);   
print(f"size: {resized_cropped_region.shape}")
```

# Annotating Images

```python
image = cv2.imread('../data/Apollo-8-Launch.jpg', cv2.IMREAD_COLOR)
plt.imshow(image[:,:,::-1]); plt.title('Original Image')
                   
```

```python
# Colors for annotation
yellow = (0, 255, 255)
red = (0, 0, 255)
magenta = (255, 0, 255)
green = (0, 255, 0)
```

## Drawing a line

```python
img_copy = image.copy()

img_copy =  cv2.line(img_copy, (0,445), (450,465), red, thickness=2, lineType=cv2.LINE_AA)

plt.imshow(img_copy[:,:,::-1]); plt.title('Line')
```

## Drawing a Circle

```python
img_copy = image.copy()

img_copy = cv2.circle(img_copy, (195, 55), 30, yellow, thickness=2, lineType=cv2.LINE_AA)

plt.imshow(img_copy[:,:,::-1]); plt.title('Circle') 
```

## Drawing a Rectangle

```python
img_copy = image.copy()

img_copy = cv2.rectangle(img_copy, (300, 150), (480, 430), magenta, thickness=2, lineType=cv2.LINE_AA)

plt.imshow(img_copy[:,:,::-1]); plt.title('Rectangle')
```

## Adding Text

```python
img_text = image.copy()

text = "Apollo 8 Launch"
font_face = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 1.5
font_color = green
font_thickness = 2

img_text = cv2.putText(img_text, text, (10, 50), font_face, font_scale, font_color, font_thickness, cv2.LINE_AA)    
plt.imshow(img_text[:,:,::-1]); plt.title('Text')
```
